# MiniMax-H3：FastH3 L4 或原版 DiffSynth

另有獨立 [ComfyUI workspace notebook](ComfyUI_Colab.ipynb) 與[操作說明](../doc/COMFYUI_WORKSPACE.md)，提供 H3 Continuum／實驗性原生 Ref2VA-VSA 的安裝、UI 與 API 入口；預設不下載模型、不生成，尚未做 GPU 驗收。

預設 **FastH3 | ComfyUI L4 (T2VA)**，請手動選 Colab **L4 GPU、標準 RAM**。FastH3 使用獨立文字／音效提示，預設 5 秒、864×480、24 fps；不需要外部素材，不自動加入 Turbo 或其他 speed LoRA。`FAST_H3_LORAS_JSON` 可選填相容 adapter。A 區啟動一次本機 ComfyUI；B 區沿用 server 與磁碟快取，ComfyUI 依需求載入／offload，並不保證全部權重常駐 VRAM。

選 **DiffSynth | Original H3 (FL2VA / Ref2VA)** 可使用下列既有原版功能。切換引擎請先重新啟動 runtime；更換 adapter 後重新執行 A 區。FastH3 不會執行原版安裝、下載或載入。

**第一次執行 A 區（1–5）；之後每片只執行 B 區（6–7）。** FL2VA 延用首尾幀／Turbo 流程；Ref2VA 可輸入多張參考圖與多段參考影片。切換模式或模型設定後，重新執行 A 區。

- **FL2VA**：文字、首圖、尾圖或接續前片；可用 4／8-step Turbo，適合快速產片與鎖定片頭片尾。
- **Ref2VA**：最多 9 張圖片、3 段影片、合計最多 12 個參考檔；可選原版 50-step 或獨立的 Ref2VA 8-step Turbo v1.0 768p LoRA。影片參考會取畫面並轉成 24 fps，不會使用影片音軌。
- 參考素材依 `REFERENCE_PATHS` 陣列順序送入模型。Ref2VA 參考圖／影片用來提供角色、場景、外觀或動作線索，不會像 FL2VA 首尾幀一樣鎖定輸出的特定時間點；超過本次模型影格長度的參考影片會由本地流程裁切。Colab 請選 A100、標準 RAM；素材數量與影片長度會增加記憶體和推論時間，40GB 顯存不保證最大組合都能執行。

建議：**8-step Turbo** 做 FL2VA 日常產片；**4-step v1.2** 快速試構圖／動作；**原版 50-step** 做品質對照或 Ref2VA。4-step 有音訊、細節與高速動作退化的社群回報，不能保證比原版好。

| 權重選單 | 建議 GPU 容量 | 約略下載量（不含 Turbo LoRA 1.29 GiB） |
| :--- | :--- | :--- |
| BF16 original | **80GB 建議**；原版 15s/480p 已實測 | 134.14 GiB |
| INT8 ConvRot pruned | **40GB 起評估／80GB 更有餘裕**；未實測此 workload | 55.08 GiB |
| NF4 pruned | **40GB 起評估**；較省容量，未實測此 workload | 25.81 GiB |
| NF4 original | **40GB 起評估**；保留未剪枝架構，未實測此 workload | 32.03 GiB |

以上是規劃建議，並非模型的硬性最低容量；offload、片長、解析度及啟用的 LoRA 都影響峰值。下載大小不等於 VRAM。V100 是 16/32GB，40/80GB 是 A100；本 notebook 的 BF16 運算需要原生 BF16 GPU，**不支援 V100**。

Ref2VA 多參考與 8-step Turbo 路徑依固定版 DiffSynth／LightX2V 權重接入並完成 notebook 靜態檢查；本工作選用 A100 80GB high-RAM runtime。已完成影片與執行紀錄保留在專案中。

````python
# @title 5. 載入模型一次（下一片沿用，不重建）
if using_fast_h3():
    get_fast_h3_runtime().ensure_ready(loras=parse_fast_h3_loras())
    print(json.dumps(get_fast_h3_runtime().status(), indent=2, ensure_ascii=False))

if not using_fast_h3():
    import gc


def adapt_comfy_qkv_lora(state_dict, head_dim=128):
    # Native H3: [head, QKV, channel]; Comfy pruned H3: [QKV, head, channel].
    # Permute B's output rows only: P(B @ A) == (P B) @ A.
    result = dict(state_dict)
    for name, weight in state_dict.items():
        if name.endswith(".attn.qkv_proj.lora_B.weight"):
            rows, rank = weight.shape
            assert rows % (3 * head_dim) == 0, f"Unexpected QKV LoRA shape: {weight.shape}"
            result[name] = weight.reshape(-1, 3, head_dim, rank).permute(1, 0, 2, 3).reshape(rows, rank).contiguous()
    return result


def configure_lora_layout(pipeline, precision):
    if precision not in ("nf4_pruned", "int8_pruned"):
        return
    from diffsynth.utils.lora.minimax_h3 import MiniMaxH3LoRALoader
    class ComfyLayoutH3LoRALoader(MiniMaxH3LoRALoader):
        def convert_state_dict(self, state_dict, suffix=".weight"):
            native_lightx2v = self.is_lightx2v_format(state_dict)
            converted = super().convert_state_dict(state_dict, suffix=suffix)
            # Already converted/Comfy LoRAs must not be permuted a second time.
            if native_lightx2v:
                assert suffix == ".weight"
                converted = adapt_comfy_qkv_lora(converted)
            return converted
    pipeline.lora_loader = ComfyLayoutH3LoRALoader


def ensure_model_loaded():
    global pipe, _LOADED_KEY, _LOADED_CONFIG, LOAD_SECONDS
    config = selected_config()
    key = config_key(config) + f"|vram={WEIGHT_VRAM_GIB}"
    if globals().get("_LOADED_KEY") == key and globals().get("pipe") is not None:
        print("沿用已載入的模型；沒有重新載入或重複加入 LoRA。")
        return pipe
    assert globals().get("_DOWNLOAD_KEY") == config_key(config), "模型設定有變更，請先執行 A 區下載。"
    _LOADED_KEY = None
    pipe = None
    gc.collect()
    torch.cuda.empty_cache()
    started = time.perf_counter()
    vram_config = {
        "offload_dtype": "disk", "offload_device": "disk",
        "onload_dtype": "disk", "onload_device": "disk",
        "preparing_dtype": torch.bfloat16, "preparing_device": "cuda",
        "computation_dtype": torch.bfloat16, "computation_device": "cuda",
    }
    configs = []
    for repo, pattern in component_specs(config):
        paths = sorted(str(repo_dir(repo) / item["file"]) for item in manifest
                       if item["repo"] == repo and fnmatch.fnmatchcase(item["file"], pattern))
        assert paths and all(Path(path).is_file() for path in paths), f"模型檔案缺少：{pattern}"
        configs.append(ModelConfig(path=paths if "*" in pattern else paths[0], **vram_config))
    try:
        processor_dir = repo_dir("MiniMaxAI/MiniMax-H3") / TASK_PROCESSOR_DIRS[config["task_mode"]] / "processor"
        pipe = MiniMaxH3Pipeline.from_pretrained(
            torch_dtype=torch.bfloat16, device="cuda", model_configs=configs,
            processor_config=ModelConfig(path=str(processor_dir)),
            vram_limit=WEIGHT_VRAM_GIB)
        if config["lora"]:
            configure_lora_layout(pipe, config["precision"])
            lora_path = repo_dir("lightx2v/Minimax-h3-Turbo") / config["lora"]
            # Preserve training alpha/rank from the pinned native header (user strength stays 1).
            pipe.load_lora(pipe.dit, ModelConfig(path=str(lora_path)), alpha=config["lora_scale"], hotload=True)
            # All three pinned native adapters have 52 attention/MLP groups, four fused targets each.
            patched = sum(bool(getattr(module, "lora_A_weights", [])) for module in pipe.dit.modules())
            assert patched == 208, f"LoRA 只套用了 {patched}/208 層，停止以免誤用未加速的模型。"
        _LOADED_CONFIG = dict(config)
        _LOADED_KEY = key
        LOAD_SECONDS = round(time.perf_counter() - started, 2)
        (SETUP_DIR / "loaded_model.json").write_text(json.dumps(
            {"config": config, "load_seconds": LOAD_SECONDS, "weight_vram_gib": WEIGHT_VRAM_GIB}, indent=2, ensure_ascii=False))
        print(f"{config['task_mode'].upper()} 模型已就緒（{LOAD_SECONDS}s）。接下來只需執行 6–7。")
        return pipe
    except Exception:
        pipe = None
        gc.collect()
        torch.cuda.empty_cache()
        raise


if not using_fast_h3():


    ensure_model_loaded()
````

- **Ref2VA**：將要參考的圖片／影片路徑按語意順序放進 `REFERENCE_PATHS` JSON 陣列，例如 `["/content/hamster.png", "/content/motion.mp4"]`。最多 9 張圖片、3 段影片、總數 12 個；每段影片 2–15 秒，全部參考影片總長不超過 15 秒。支援 PNG/JPG/JPEG/WEBP 與 MP4/MOV。
- Ref2VA 可選 `Ref2VA Turbo 8-step v1.0 768p` 或 `Original 50-step`，並搭配 `NF4 pruned`／`NF4 original`。8-step LoRA 專屬 Ref2VA；A 區會下載對應的 Ref2VA 權重與 adapter。請在 `PROMPT` 寫清楚 `<Picture 1>`、`<Video 1>` 等素材各自要提供什麼參考。
- 影片參考會抽取並整理為 24 fps，送入模型的是畫面；參考影片的音軌不會被使用。超過本次模型影格長度的影片會裁切後送入模型。
- 每次建立獨立時間戳資料夾，同一個 `RUN_NAME` 也不會覆蓋前片。產片前請先將素材上傳至 Colab Files。
- `DURATION_SECONDS` 可調 5–15 秒，每秒一格，預設 5 秒。較長片長會增加 VRAM 需求；768p／15s 對 80GB 也不是保證。
- `SEED` 預設為 `random`，每次產片抽新 seed；輸入整數可固定 seed，實際使用值會記在 `metrics.json`。
- 768p 模型名稱代表訓練版本；下方仍可選 480p 輸出。

````python
# @title 7. 產片、驗證與預覽（沿用模型）
IMAGE_EXTENSIONS = {".png", ".jpg", ".jpeg", ".webp"}
VIDEO_EXTENSIONS = {".mp4", ".mov"}

def generate_fast_h3_clip():
    global OUTPUT_PATH, RUN_DIR, FAST_H3_RESULT
    runtime = get_fast_h3_runtime()
    adapters = parse_fast_h3_loras()
    runtime.ensure_ready(loras=adapters)
    FAST_H3_RESULT = runtime.generate(
        prompt=FAST_H3_PROMPT, width=FAST_H3_WIDTH, height=FAST_H3_HEIGHT,
        duration_seconds=FAST_H3_DURATION_SECONDS, seed=FAST_H3_SEED,
        loras=adapters, run_name=FAST_H3_RUN_NAME, timeout=1800,
        generation_mode=FAST_H3_MODE, reference_paths=parse_fast_h3_references())
    OUTPUT_PATH = Path(FAST_H3_RESULT["video_path"])
    RUN_DIR = Path(FAST_H3_RESULT["run_dir"])
    print(json.dumps(FAST_H3_RESULT, indent=2, ensure_ascii=False))
    from IPython.display import Video, display
    display(Video(str(OUTPUT_PATH), embed=True, width=720))
    return OUTPUT_PATH

if using_fast_h3():
    generate_fast_h3_clip()

if not using_fast_h3():
    import gc, threading, traceback
    from datetime import datetime, timezone
    import av, numpy as np
    from PIL import Image
    from IPython.display import Video, display


def _reference_image_size(width, height):
    scale = min(768 / min(width, height), math.sqrt((768 * 1344) / (width * height)))
    out_width = max(32, int(round(width * scale / 32)) * 32)
    out_height = max(32, int(round(height * scale / 32)) * 32)
    return out_width, out_height


def read_reference_video(path, max_frames):
    frames = []
    with av.open(str(path)) as container:
        if not container.streams.video:
            raise ValueError(f"參考影片沒有影像串流：{path}")
        stream = container.streams.video[0]
        rate = stream.average_rate or stream.base_rate
        if rate is None or float(rate) <= 0:
            raise ValueError(f"無法讀取參考影片影格率：{path}")
        source_fps = float(rate)
        if stream.duration is not None and stream.time_base is not None:
            duration = float(stream.duration * stream.time_base)
        elif stream.frames and stream.frames > 0:
            duration = float(stream.frames) / source_fps
        elif container.duration is not None:
            duration = float(container.duration) / 1_000_000
        else:
            raise ValueError(f"無法讀取參考影片長度：{path}")
        if not 2 <= duration <= 15:
            raise ValueError(f"參考影片需為 2–15 秒：{path}（{duration:.2f}s）")
        target_count = min(max_frames, max(1, round(duration * 24)))
        source_count = int(stream.frames) if stream.frames and stream.frames > 0 else max(1, round(duration * source_fps))
        source_indices = [min(source_count - 1, max(0, round(k * source_fps / 24))) for k in range(target_count)]
        targets_by_source = {}
        for target_index, source_index in enumerate(source_indices):
            targets_by_source.setdefault(source_index, []).append(target_index)
        output_frames = [None] * target_count
        last_source_frame = None
        max_source_index = max(targets_by_source)
        for source_index, frame in enumerate(container.decode(video=0)):
            if source_index > max_source_index:
                break
            targets = targets_by_source.get(source_index)
            if targets:
                source_image = frame.to_image().convert("RGB")
                resized = source_image.resize(_reference_image_size(*source_image.size), Image.Resampling.LANCZOS)
                last_source_frame = resized
                for target_index in targets:
                    output_frames[target_index] = resized.copy()
        if last_source_frame is None:
            raise ValueError(f"參考影片沒有可解碼影格：{path}")
        for target_index, frame in enumerate(output_frames):
            if frame is None:
                output_frames[target_index] = last_source_frame.copy()
    return output_frames, duration, source_fps


def prepare_clip():
    config = selected_config()
    expected_key = config_key(config) + f"|vram={WEIGHT_VRAM_GIB}"
    assert globals().get("pipe") is not None and globals().get("_LOADED_KEY") == expected_key, (
        "尚未載入所選模型，或 A 區選項已變更。請先完成 A 區；本格不會自行重載模型。")
    assert re.fullmatch(r"[A-Za-z0-9][A-Za-z0-9_-]{0,63}", RUN_NAME), "RUN_NAME 請用英數、底線或連字號。"
    assert isinstance(PROMPT, str) and PROMPT.strip(), "Prompt 不可為空。"
    assert isinstance(DURATION_SECONDS, int) and 5 <= DURATION_SECONDS <= 15, "片長請選擇 5 至 15 秒的整數。"
    seed_setting = str(SEED).strip().lower()
    assert seed_setting == "random" or (seed_setting.isdecimal() and 0 <= int(seed_setting) < 2**32), (
        "SEED 請填 random，或填 0 到 2^32-1 的整數。")
    sizes = {"480p landscape (832x480)": (480, 832), "480p portrait (480x832)": (832, 480),
             "720p-adj landscape (1312x736)": (736, 1312),
             "768p landscape (1344x768)": (768, 1344), "768p portrait (768x1344)": (1344, 768)}
    assert RESOLUTION in sizes
    height, width = sizes[RESOLUTION]
    output_frames = DURATION_SECONDS * 24
    model_frames = 5 + 17 * math.ceil((output_frames - 5) / 17)
    reference_paths = json.loads(REFERENCE_PATHS)
    assert isinstance(reference_paths, list) and all(isinstance(value, str) and value.strip() for value in reference_paths), (
        'REFERENCE_PATHS must be a JSON array of file paths, e.g. ["/content/character.png", "/content/motion.mp4"].')
    first_image_meta = last_image_meta = None
    if config["task_mode"] == "ref2va":
        assert FIRST_IMAGE_PATH.strip() == "" and LAST_IMAGE_PATH.strip() == "" and not CONTINUE_PREVIOUS, (
            "Ref2VA 使用 REFERENCE_PATHS；請清空 FIRST_IMAGE_PATH／LAST_IMAGE_PATH 並關閉 CONTINUE_PREVIOUS。")
        assert reference_paths, "Ref2VA 至少需要一張參考圖片或一段參考影片。"
        assert len(reference_paths) <= 12, "Ref2VA 最多接受 12 個參考檔。"
        references, reference_manifest = [], []
        image_count = video_count = 0
        total_video_seconds = 0.0
        for raw_path in reference_paths:
            path = Path(raw_path).expanduser()
            if not path.is_file():
                raise FileNotFoundError(f"請先上傳參考素材到 Colab：{path}")
            suffix = path.suffix.lower()
            if suffix in IMAGE_EXTENSIONS:
                if image_count >= 9:
                    raise ValueError("Ref2VA 最多接受 9 張參考圖片。")
                with Image.open(path) as source:
                    references.append({"type": "image", "image": source.convert("RGB")})
                image_count += 1
                reference_manifest.append({"type": "image", "path": str(path)})
            elif suffix in VIDEO_EXTENSIONS:
                if video_count >= 3:
                    raise ValueError("Ref2VA 最多接受 3 段參考影片。")
                frames, video_seconds, source_fps = read_reference_video(path, model_frames)
                references.append({"type": "video", "video": frames})
                video_count += 1
                total_video_seconds += video_seconds
                reference_manifest.append({"type": "video", "path": str(path), "duration_seconds": round(video_seconds, 3),
                                          "duration_used_seconds": round(len(frames) / 24, 3),
                                          "source_fps": round(source_fps, 3), "frames_at_24fps": len(frames)})
            else:
                raise ValueError(f"不支援的參考素材格式：{path.suffix}；請使用 PNG/JPG/JPEG/WEBP 或 MP4/MOV。")
        if total_video_seconds > 15:
            raise ValueError(f"參考影片總長不可超過 15 秒（目前 {total_video_seconds:.2f}s）。")
        reference_kwargs = {"references": references, "ref_image_short_edge": max(256, int(round((min(height, width) + 96) / 32) * 32))}
    else:
        assert not reference_paths, "FL2VA 請將 REFERENCE_PATHS 留空（[]），並使用首圖／尾圖欄位。"
        first_path = FIRST_IMAGE_PATH.strip()
        if CONTINUE_PREVIOUS:
            assert not first_path, "續接前片時請將 FIRST_IMAGE_PATH 留空，避免兩個首圖來源。"
            assert globals().get("LAST_VERIFIED_END_FRAME"), "本 kernel 還沒有成功驗證的前片；也可以自行上傳前片最後一格。"
            first_path = str(LAST_VERIFIED_END_FRAME)
        images, indices = [], []
        for label, filename, index in (("first", first_path, 0), ("last", LAST_IMAGE_PATH.strip(), -1)):
            if not filename:
                continue
            path = Path(filename).expanduser()
            if not path.is_file():
                raise FileNotFoundError(f"請先上傳 {label} 圖片到 Colab：{path}")
            with Image.open(path) as source:
                images.append(source.convert("RGB"))
            indices.append(index)
        reference_kwargs = {"keyframes": images, "keyframe_indices": indices} if images else {}
        first_image_meta = first_path or None
        last_image_meta = LAST_IMAGE_PATH.strip() or None
        reference_manifest = ([{"type": "first_frame", "path": str(first_path)}] if first_path else [])
        if LAST_IMAGE_PATH.strip():
            reference_manifest.append({"type": "last_frame", "path": LAST_IMAGE_PATH.strip()})
    return dict(config=config, height=height, width=width, output_frames=output_frames,
                model_frames=model_frames, reference_kwargs=reference_kwargs,
                reference_manifest=reference_manifest, first_image=first_image_meta, last_image=last_image_meta)


def generate_clip():
    global OUTPUT_PATH, RUN_DIR, LAST_VERIFIED_END_FRAME, LAST_VERIFIED_OUTPUT
    clip = prepare_clip()  # Validate all new media before allocating GPU work/output folders.
    config = clip["config"]
    seed_setting = str(SEED).strip().lower()
    if seed_setting == "random":
        import secrets
        run_seed = secrets.randbits(32)
    else:
        run_seed = int(seed_setting)
    stamp = datetime.now(timezone.utc).strftime("%Y%m%dT%H%M%S_%fZ")
    RUN_DIR = RUNS_ROOT / f"{RUN_NAME}_{stamp}"
    RUN_DIR.mkdir(parents=True, exist_ok=False)
    OUTPUT_PATH = RUN_DIR / "video.mp4"
    (RUN_DIR / "prompt.txt").write_text(PROMPT)
    (RUN_DIR / "model_manifest.json").write_text(json.dumps(manifest, indent=2))
    (RUN_DIR / "hardware.json").write_text(json.dumps(hardware, indent=2))
    gc.collect()
    torch.cuda.empty_cache()
    torch.cuda.reset_peak_memory_stats()
    started = time.perf_counter()
    phase = "generating"
    samples, monitor_errors = [], []
    stop_monitor = threading.Event()
    process = psutil.Process()
    def sample_resources():
        entry = {"seconds": round(time.perf_counter() - started, 2), "phase": phase,
                 "rss_gib": round(process.memory_info().rss / 2**30, 3),
                 "ram_available_gib": round(psutil.virtual_memory().available / 2**30, 3)}
        try:
            values = subprocess.check_output(["nvidia-smi", "--id=0", "--query-gpu=memory.used,utilization.gpu",
                                              "--format=csv,noheader,nounits"], text=True, timeout=5).strip().split(",")
            entry["gpu_used_gib"] = round(float(values[0]) / 1024, 3)
            entry["gpu_utilization_percent"] = float(values[1])
        except (subprocess.SubprocessError, ValueError):
            pass
        samples.append(entry)
        (RUN_DIR / "progress.json").write_text(json.dumps(entry))
    def monitor_resources():
        while not stop_monitor.is_set():
            try:
                sample_resources()
            except Exception as exc:
                monitor_errors.append(str(exc))
                return
            stop_monitor.wait(3)
    metrics = {"status": "running", "started_utc": stamp, "run_name": RUN_NAME, "prompt": PROMPT,
               "model": config, "task_mode": config["task_mode"], "references": clip["reference_manifest"],
               "precision": config["precision"], "steps": config["steps"],
               "seed": run_seed, "seed_mode": "random" if seed_setting == "random" else "fixed",
               "first_image": clip["first_image"], "last_image": clip["last_image"],
               "width": clip["width"], "height": clip["height"], "frames": clip["model_frames"],
               "output_frames": clip["output_frames"], "requested_duration_seconds": DURATION_SECONDS,
               "hardware": hardware, "versions": versions, "diffsynth_commit": DIFFSYNTH_COMMIT,
               "model_revisions": REVISIONS, "offload": "disk", "weight_vram_budget_gib": WEIGHT_VRAM_GIB,
               "setup_model_load_seconds": LOAD_SECONDS, "pipeline_reused": True}
    monitor = threading.Thread(target=monitor_resources, daemon=True)
    monitor.start()
    video = audio = None
    try:
        print(f"沿用 {config['task_mode'].upper()} / {config['id']} / {config['precision']}，{config['steps']} steps，seed={run_seed}；輸出：{RUN_DIR}", flush=True)
        with torch.inference_mode():
            video, audio = pipe(prompt=PROMPT, height=clip["height"], width=clip["width"],
                num_frames=clip["model_frames"], num_inference_steps=config["steps"], seed=run_seed,
                cfg_scale=1.0, flow_shift=config["video_shift"], audio_flow_shift=config["audio_shift"],
                **clip["reference_kwargs"])
        torch.cuda.synchronize()
        metrics["generation_seconds"] = round(time.perf_counter() - started, 2)
        phase = "encoding"
        assert len(video) >= clip["output_frames"] and audio.shape[-1] >= DURATION_SECONDS * 32000
        video, audio = video[:clip["output_frames"]], audio[..., :DURATION_SECONDS * 32000]
        write_video_audio(video=video, audio=audio, output_path=str(OUTPUT_PATH), fps=24, audio_sample_rate=32000)
        video = audio = None
        phase = "verifying"
        with av.open(str(OUTPUT_PATH)) as container:
            assert container.streams.video and container.streams.audio, "MP4 缺少影像或音訊。"
            stream = container.streams.video[0]
            video_info = {"width": stream.width, "height": stream.height,
                          "fps": float(stream.average_rate), "codec": stream.codec_context.name,
                          "duration_seconds": float(stream.duration * stream.time_base)}
            decoded_frames, last_frame = 0, None
            for frame in container.decode(video=0):
                if decoded_frames in (0, clip["output_frames"] // 2):
                    label = "first" if decoded_frames == 0 else "middle"
                    frame.to_image().save(RUN_DIR / f"preview_{label}.png")
                last_frame = frame
                decoded_frames += 1
            assert last_frame is not None
            last_frame.to_image().save(RUN_DIR / "preview_last.png")
        with av.open(str(OUTPUT_PATH)) as container:
            stream = container.streams.audio[0]
            audio_info = {"sample_rate": stream.rate, "channels": len(stream.layout.channels),
                          "codec": stream.codec_context.name, "duration_seconds": float(stream.duration * stream.time_base)}
            square_sum, count = 0.0, 0
            for frame in container.decode(audio=0):
                chunk = frame.to_ndarray().astype(np.float64)
                assert np.isfinite(chunk).all(), "音訊含有非有限值。"
                square_sum += float(np.square(chunk).sum())
                count += chunk.size
            assert count > 0
            rms = float(np.sqrt(square_sum / count))
        assert (video_info["width"], video_info["height"]) == (clip["width"], clip["height"])
        assert decoded_frames == clip["output_frames"] and video_info["fps"] == 24
        assert abs(video_info["duration_seconds"] - DURATION_SECONDS) < 1 / 24
        assert audio_info["sample_rate"] == 32000 and audio_info["channels"] == 2
        assert abs(audio_info["duration_seconds"] - DURATION_SECONDS) < 1 / 24
        assert rms > 1e-7, "音訊為靜音；請檢查結果。"
        metrics.update(status="verified", decoded_frames=decoded_frames, video=video_info, audio=audio_info,
                       decoded_audio_rms=rms, output_file=str(OUTPUT_PATH), output_bytes=OUTPUT_PATH.stat().st_size)
        LAST_VERIFIED_END_FRAME = RUN_DIR / "preview_last.png"
        LAST_VERIFIED_OUTPUT = OUTPUT_PATH
        (RUNS_ROOT / "latest.json").write_text(json.dumps({"video": str(OUTPUT_PATH), "run_dir": str(RUN_DIR),
                                                         "last_frame": str(LAST_VERIFIED_END_FRAME)}))
        phase = "verified"
    except Exception as exc:
        metrics.update(status="failed", error=f"{type(exc).__name__}: {exc}", traceback=traceback.format_exc())
        phase = "failed"
        raise
    finally:
        stop_monitor.set()
        monitor.join(timeout=10)
        try:
            sample_resources()
        except Exception as exc:
            monitor_errors.append(str(exc))
        metrics.update(total_seconds=round(time.perf_counter() - started, 2),
                       peak_torch_allocated_gib=round(torch.cuda.max_memory_allocated() / 2**30, 3),
                       peak_torch_reserved_gib=round(torch.cuda.max_memory_reserved() / 2**30, 3),
                       peak_gpu_used_gib_sampled=max((s.get("gpu_used_gib", 0) for s in samples), default=0),
                       peak_process_rss_gib_sampled=max((s["rss_gib"] for s in samples), default=0),
                       monitor_errors=monitor_errors)
        (RUN_DIR / "metrics.json").write_text(json.dumps(metrics, indent=2, ensure_ascii=False))
        (RUN_DIR / "resources.json").write_text(json.dumps(samples, indent=2))
        video = audio = None
        gc.collect()
        print(json.dumps(metrics, indent=2, ensure_ascii=False), flush=True)
    print(f"PASS：{decoded_frames} frames，{DURATION_SECONDS}s，32 kHz stereo。下一片只需修改 6，再執行 7。")
    display(Video(str(OUTPUT_PATH), embed=True, width=720))
    return OUTPUT_PATH


if not using_fast_h3():


    generate_clip()
````
